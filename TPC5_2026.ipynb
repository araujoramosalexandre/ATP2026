```python
turma = []

def menu ():
    print("(1) Criar uma turma")
    print("(2) Inserir um aluno na turma")
    print("(3) Listar uma turma")
    print("(4) Consultar um aluno por id")
    print("(5) Guardar a turma em ficheiro")
    print("(6) Carregar uma turma de um ficheiro")
    print("(0) Sair da aplicação")

def CriarTurma():
    letra = input("Qual a letra que deseja associar a esta turma")
    print(f"turma {letra}: {turma}")

def InserirAluno():
    nome = input("Qual é o nome do aluno?")
    id = input("Qual o id do aluno?")
    notaTPC = input(f"Qual é a nota do TPC do {nome}")
    notaProj = input(f"Qual é a nota do Projeto do {nome}")
    notaTeste = input(f"Qual é a nota do Teste do {nome}")
    aluno = (nome, id, [notaTPC, notaProj, notaTeste])
    turma.append(aluno)
    print(turma)

def ListarTurma():
    print(turma)

def ConsultarId():
    i = 0
    ID = input("Qual é o id do aluno?")
    while i < len(turma):
        if ID == turma[i][1]:
            print(turma[i])
        i = i + 1

def GuardarTurma(turma, nomeficheiro):
    file = open(nomeficheiro,"w")
    res = ""
    i = 0
    while i < len(turma):
        res = res + turma[i][0] + ";" + turma[i][1] + ";" + turma[i][2][0] + "," + turma[i][2][1] + "," + turma[i][2][2] + "|"
        res = res + "\n"
        i = i + 1
    file.write(res)
    file.close()

def CarregarTurma(nomeficheiro):
    file = open(nomeficheiro,"r")
    text = file.read()
    c = text.split("\n")
    aluno = []
    print(c)
    for d in c[:-1]:
        e = d.split("|")
        print(e)
        for f in e[:-1]:
            g = f.split("j")
            print(g)
            aluno.append(g)
        turma.append(aluno) 
    return turma
    print(CarregarTurma("turma.txt"))


menu()
while True:
    a = int(input("Selecione uma opção do menu"))
    if a == 1:
        CriarTurma()

    if a == 2:
        InserirAluno()

    if a == 3:
        ListarTurma()
    
    if a == 4:
        ConsultarId()
    
    if a == 5:
        GuardarTurma(turma, "turma.txt")

    if a == 6:
        CarregarTurma("turma.txt")

    if a == 0:
        break
```
