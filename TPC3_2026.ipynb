```python
def jogar_computador_primeiro():
    print("\nComputador Joga Primeiro")
    total = 0
    print(f"Total inicial: {total}")
    
    jogada_pc = 1
    total = total + jogada_pc
    print(f"O computador jogou: {jogada_pc}. Total atual: {total}")
    
    while total < 100:
        jogada_jogador = int(input("Escolhe um número de 1 a 10: "))      
        total = total + jogada_jogador
        print(f"Total atual: {total}")

        if total == 100:
            print("Parabéns! Ganhaste o jogo.")
        elif total > 100:
            print("Ultrapassaste os 100! O computador venceu.")
        else:
            if total >= 90:
                jogada_pc = 100 - total
                total = total + jogada_pc
                print(f"O computador jogou: {jogada_pc}. Total atual: {total}")
                print("O computador venceu!")


            else:
                jogada_pc = 11 - jogada_jogador
                total = total + jogada_pc
                print(f"O computador jogou: {jogada_pc}. Total atual: {total}")

                if total == 100:
                    print("O computador venceu!")

            

def jogar_computador_segundo():
    print("\nComputador Joga em Segundo")
    total = 0
    print(f"Total inicial: {total}")
    
    while total < 100:
        jogada_jogador = int(input("Escolhe um número de 1 a 10: "))   
        total = total + jogada_jogador
        print(f"Total atual: {total}")
        
        if total == 100:
            print("Parabéns! Ganhaste o jogo.")
        elif total > 100:
            print("Ultrapassaste os 100! O computador venceu.")
        else:
            if total > 90:
                jogada_pc = 100 - total
                total = total + jogada_pc
                print(f"O computador jogou: {jogada_pc}. Total atual: {total}")


            else:
                jogada_pc = 11 - jogada_jogador
                total = total + jogada_pc
                print(f"O computador jogou: {jogada_pc}. Total atual: {total}")
            
def main():
    print("Corrida para o 100")
    print("(1) Computador joga primeiro")
    print("(2) Computador joga em segundo")
    
    opcao = input("Escolhe (1 ou 2): ")
    
    if opcao == '1':
        jogar_computador_primeiro()
    elif opcao == '2':
        jogar_computador_segundo()
    else:
        print("Opção inválida. Execute o programa novamente.")

if __name__ == "__main__":
    main()
```
