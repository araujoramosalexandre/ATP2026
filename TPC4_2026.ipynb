```python
import random

lista = []

menu = -1

def soma(lista):
    soma = 0
    for elem in lista:
        soma = soma + elem
    return soma

def media(lista):
    return soma(lista)/len(lista)

def maior(lista):
    maior = lista[0]
    i = 1
    while i < len(lista):
        if lista[i] > maior:
            maior = lista[i]
            i = i + 1
        else:
            i = i + 1
    return maior

def menor (lista):
    menor = lista[0]
    i = 1
    while i < len(lista):
        if lista[i] < menor:
            menor = lista[i]
            i = i + 1
        else:
            i = i + 1
    return menor

def estaOrdenadaCrescente(lista):
    ordenada = "Não"
    lista_crescente = sorted(lista)
    if lista_crescente == lista:
        ordenada = "Sim"
        return ordenada
    else:
        return ordenada

def estaOrdenadaDecrescente(lista):
    ordenada = "Não"
    lista_crescente = sorted(lista)
    lista_decrescente = lista_crescente[::-1]
    if lista_decrescente == lista:
        ordenada = "Sim"
        return ordenada
    else:
        return ordenada

def posicao_lista(lista, elem):
    i = 0
    if elem in lista:
        while i < len(lista):
            if lista[i] == elem:
                print(f"O elemnto {elem} está na posição {i+1} da lista (indice {i})")
                i = i + 1
            else:
                i = i + 1
    else:
        print(-1)
    
        


while menu != 0:
    menu = int(input("""
(1) Criar Lista
(2) Ler Lista
(3) Soma
(4) Média
(5) Maior
(6) Menor
(7) estaOrdenada por ordem crescente
(8) estaOrdenada por ordem decrescente
(9) Procura um elemento
(0) Sair
"""))

    if menu == 1:
        lista_random = []
        for i in range(10):
            numero = random.randint(1,100)
            lista_random.append(numero)
        lista = lista_random
        print(lista)

    elif menu == 2:
        b = int(input("Quantos números deseja ler?"))
        lista_lida = []
        i = 1
        while i <= b:
            c = int(input(f"Introduza um número {i}/{b}"))
            lista_lida.append(c)
            i = i + 1
        lista = lista_lida
        print(lista)

    elif menu == 3:
        print(soma(lista))

    elif menu == 4:
        print(media(lista))
    
    elif menu == 5:
        print(maior(lista))

    elif menu == 6:
        print(menor(lista))

    elif menu == 7:
        print(estaOrdenadaCrescente(lista))

    elif menu == 8:
        print(estaOrdenadaDecrescente(lista))

    elif menu == 9:
        d = int(input("Que número deseja procurar na lista?"))
        print(posicao_lista(lista, d))
```
